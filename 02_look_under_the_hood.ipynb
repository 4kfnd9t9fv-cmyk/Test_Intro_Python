## 1. Load and prepare a training text

```python
from collections import defaultdict, Counter
from pathlib import Path
import random

text = Path("data/fairy_tales.txt").read_text(encoding="utf-8").lower()
print(text[:250])
```

## 2. What does the training function do?
The two characters in `zip(text, text[1:])` move through the text as neighbouring pairs.

**Mini-task:** Before running the next cell, predict what the first five pairs will be.

```python
first_pairs = list(zip(text, text[1:]))[:5]
print(first_pairs)

def train_character_model(training_text):
    transitions = defaultdict(Counter)
    for current_char, next_char in zip(training_text, training_text[1:]):
        transitions[current_char][next_char] += 1
    return transitions

model = train_character_model(text)
print("Model trained.")
```

## 3. Explore what the model learned
Change `letter` to another character. What changes?

```python
letter = "t"
print("After", repr(letter), "the model observed:")
print(model[letter].most_common(10))
```

## 4. Make a small code change
The function below generates text. Change the default length from 200 to 100, then run the cell. Next, try 500.

**Challenge:** Can you explain why the output is not exactly the same every time?

```python
def generate_text(start, transitions, length=100):
    result = start.lower()
    current = result[-1]
    for _ in range(length):
        choices = transitions.get(current)
        if not choices:
            current = random.choice(list(transitions.keys()))
            result += current
            continue
        chars = list(choices.keys())
        counts = list(choices.values())
        next_char = random.choices(chars, weights=counts, k=1)[0]
        result += next_char
        current = next_char
    return result

print(generate_text("once upon a time", model))
```
