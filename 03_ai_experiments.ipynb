Your task is to design a fair experiment. Change one factor at a time and observe the output.

This notebook uses the same simple character-level model as the first two notebooks.

```python
from pathlib import Path
from collections import defaultdict, Counter
import random

def train_character_model(training_text):
    transitions = defaultdict(Counter)
    for a, b in zip(training_text, training_text[1:]):
        transitions[a][b] += 1
    return transitions

def generate_text(start, transitions, length=250, randomness=0.8):
    result = start.lower()
    current = result[-1]
    for _ in range(length):
        choices = transitions.get(current)
        if not choices:
            current = random.choice(list(transitions.keys()))
            result += current
            continue
        chars = list(choices.keys())
        counts = list(choices.values())
        weights = [n ** (1 / max(randomness, 0.05)) for n in counts]
        nxt = random.choices(chars, weights=weights, k=1)[0]
        result += nxt
        current = nxt
    return result

print("Helper functions are ready.")
```

## Investigation A – Does training material matter?
Choose one dataset, train the model, and generate text. Repeat with another dataset.

**Your prediction:** What kind of text do you expect each model to produce?

```python
files = {
    "fairy tales": "data/fairy_tales.txt",
    "football": "data/football.txt",
    "science fiction": "data/science_fiction.txt"
}

chosen_dataset = "football"  # Change this
training_text = Path(files[chosen_dataset]).read_text(encoding="utf-8").lower()
model = train_character_model(training_text)

print("Dataset:", chosen_dataset)
print(generate_text("the ", model, length=300))
```

## Investigation B – Does randomness matter?
Keep the dataset and starting phrase the same. Change only `randomness`.

Try 0.2, 0.8 and 1.4. Record what happens. This setting changes the sampling distribution; it does not make the model more knowledgeable.

```python
for randomness in [0.2, 0.8, 1.4]:
    print("\nRANDOMNESS =", randomness)
    print(generate_text("the ", model, length=180, randomness=randomness))
```
